# Building energy efficiency: multi-target regression

**Question:** How accurately can building geometry predict heating and cooling loads?

Source: [UCI Energy Efficiency](https://archive.ics.uci.edu/dataset/242/energy+efficiency). This demonstrates clean multi-output modelling and error reporting in original units.

```python
from pathlib import Path
import pandas as pd
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, r2_score
from sklearn.model_selection import train_test_split

df=pd.read_excel(Path('../datasets/portfolio-expansion/energy/ENB2012_data.xlsx'))
df=df.iloc[:,:10].copy()  # Drop trailing blank spreadsheet columns.
df.columns=['relative_compactness','surface_area','wall_area','roof_area','overall_height','orientation','glazing_area','glazing_distribution','heating_load','cooling_load']
X=df.iloc[:,:8]; y=df[['heating_load','cooling_load']]
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42)
model=RandomForestRegressor(n_estimators=500,min_samples_leaf=2,random_state=42,n_jobs=-1).fit(X_train,y_train)
pred=model.predict(X_test)
for i,target in enumerate(y.columns): print(target,'MAE=',round(mean_absolute_error(y_test.iloc[:,i],pred[:,i]),2),'R²=',round(r2_score(y_test.iloc[:,i],pred[:,i]),3))
```

```python
# Geometry features that consistently matter are candidates for engineering review.
importance=pd.Series(model.feature_importances_,index=X.columns).sort_values(ascending=False)
display(importance.to_frame('importance'))
# Validation here is predictive, not causal: geometry changes should still be simulated or experimentally tested.
```
